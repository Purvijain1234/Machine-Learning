# Polynomial Regression

## Introduction

Polynomial Regression is a supervised machine learning algorithm used to predict a continuous numerical value when the relationship between the input and output is **non-linear or curved**.

It is an extension of Linear Regression that adds polynomial powers of the input features.

The general equation is:

$$
\hat{y} = \beta_0 + \beta_1x + \beta_2x^2 + \beta_3x^3 + \cdots + \beta_dx^d
$$

Where:

* $x$ = Input Feature
* $y$ = Actual Output
* $\hat{y}$ = Predicted Output
* $\beta_0$ = Intercept
* $\beta_1, \beta_2, \ldots, \beta_d$ = Model Coefficients
* $d$ = Degree of Polynomial

### Examples

* Predicting salary based on experience when salary growth is non-linear.
* Predicting house prices based on area.
* Modeling population growth.
* Modeling temperature changes.
* Predicting sales with non-linear trends.

> **The goal is to find a polynomial curve that minimizes prediction errors.**

---

# Mathematical Representation

The general Polynomial Regression equation is:

$$
\boxed{
\hat{y} = \beta_0 + \beta_1x + \beta_2x^2 + \cdots + \beta_dx^d
}
$$

### Degree 2 Polynomial

$$
\hat{y} = \beta_0 + \beta_1x + \beta_2x^2
$$

### Degree 3 Polynomial

$$
\hat{y} = \beta_0 + \beta_1x + \beta_2x^2 + \beta_3x^3
$$

Where:

* $\hat{y}$ = Predicted value
* $x$ = Input feature
* $\beta_0$ = Intercept
* $\beta_1, \beta_2, \ldots$ = Coefficients
* $d$ = Polynomial degree

---

# Connection with Linear Regression

Polynomial Regression is closely related to Linear Regression.

It first transforms the original features into polynomial features and then applies Linear Regression.

For example:

$$
X \rightarrow X, X^2, X^3, \ldots, X^d
$$

Therefore:

$$
\boxed{
\text{Polynomial Regression}
=
\text{Polynomial Features}
+
\text{Linear Regression}
}
$$

Although the relationship is non-linear with respect to $X$, the model is **linear with respect to its coefficients**.

---

# Polynomial Features

Polynomial Regression creates additional features by calculating powers of the original feature.

For example, if the original feature is:

$$
X
$$

For degree 3:

$$
X,\ X^2,\ X^3
$$

Example:

| $X$ | $X^2$ | $X^3$ |
| --: | ----: | ----: |
|   1 |     1 |     1 |
|   2 |     4 |     8 |
|   3 |     9 |    27 |
|   4 |    16 |    64 |

These transformed features are then given to Linear Regression.

---

# Polynomial Degree

The **degree** determines the complexity and flexibility of the polynomial.

## Degree 1

$$
\hat{y} = \beta_0 + \beta_1x
$$

* Straight line
* Same as Linear Regression
* Lowest complexity

---

## Degree 2

$$
\hat{y} = \beta_0 + \beta_1x + \beta_2x^2
$$

* Quadratic equation
* Produces a curved/parabolic relationship

---

## Degree 3

$$
\hat{y} =
\beta_0 +
\beta_1x +
\beta_2x^2 +
\beta_3x^3
$$

* More flexible than degree 2
* Can represent S-shaped relationships

---

## Higher Degree

$$
\hat{y} =
\beta_0 +
\beta_1x +
\beta_2x^2 +
\cdots +
\beta_dx^d
$$

As degree increases:

$$
\text{Degree} \uparrow
\Rightarrow
\text{Complexity} \uparrow
$$

Higher degree means:

* More flexibility
* More features
* Higher chance of overfitting

---

# How Polynomial Regression Works

The algorithm follows these steps:

1. Collect the dataset.
2. Split data into training and testing sets.
3. Choose the polynomial degree.
4. Transform input features into polynomial features.
5. Train Linear Regression on transformed features.
6. Make predictions.
7. Calculate prediction errors.
8. Minimize the loss function.
9. Evaluate the model.
10. Select the appropriate polynomial degree.

---

# Prediction

For a degree-2 polynomial:

$$
\hat{y} = \beta_0 + \beta_1x + \beta_2x^2
$$

Suppose:

$$
\beta_0 = 2
$$

$$
\beta_1 = 3
$$

$$
\beta_2 = 1
$$

and:

$$
x = 4
$$

Then:

$$
\hat{y}
=
2 + (3)(4) + (1)(4^2)
$$

$$
\hat{y}
=
2 + 12 + 16
$$

$$
\boxed{\hat{y} = 30}
$$

---

# Residuals (Errors)

Residuals represent the difference between actual and predicted values.

$$
\boxed{
Residual = Actual - Predicted
}
$$

Therefore:

$$
\boxed{
e_i = y_i - \hat{y}_i
}
$$

Example:

| Actual | Predicted | Residual |
| -----: | --------: | -------: |
|     10 |         8 |        2 |
|     15 |        12 |        3 |
|     20 |        18 |        2 |

Smaller residuals generally indicate better predictions.

---

# Loss Function

The loss function measures how far the predicted values are from the actual values.

The objective of Polynomial Regression is to minimize the prediction error.

$$
\boxed{
\text{Minimize Loss}
}
$$

---

# Mean Squared Error (MSE)

The commonly used loss function is Mean Squared Error.

$$
\boxed{
MSE =
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
}
$$

Where:

* $n$ = Number of observations
* $y_i$ = Actual value
* $\hat{y}_i$ = Predicted value

The objective is:

$$
\boxed{
\min MSE
}
$$

---

# Sum of Squared Errors (SSE)

SSE represents the total squared prediction error.

$$
\boxed{
SSE =
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
}
$$

MSE is calculated as:

$$
\boxed{
MSE = \frac{SSE}{n}
}
$$

---

# Why Squared Errors?

Squaring the errors is useful because:

1. Negative errors become positive.
2. Large errors receive greater penalties.
3. The function becomes differentiable.
4. Positive and negative errors cannot cancel each other.

Example:

| Error | Squared Error |
| ----: | ------------: |
|     2 |             4 |
|    -2 |             4 |
|     5 |            25 |
|    -5 |            25 |

---

# Matrix Representation

Polynomial Regression can be represented using matrices.

For a degree-2 polynomial:

$$
X =
\begin{bmatrix}
1 & x_1 & x_1^2 \\
1 & x_2 & x_2^2 \\
1 & x_3 & x_3^2 \\
\vdots & \vdots & \vdots \\
1 & x_n & x_n^2
\end{bmatrix}
$$

The coefficient vector is:

$$
\beta =
\begin{bmatrix}
\beta_0 \\
\beta_1 \\
\beta_2
\end{bmatrix}
$$

The prediction can be written as:

$$
\boxed{
\hat{y} = X\beta
}
$$

---

# Normal Equation

For ordinary least squares, the coefficients can be obtained using the Normal Equation:

$$
\boxed{
\beta = (X^TX)^{-1}X^Ty
}
$$

This finds the coefficients that minimize the squared error when the required inverse exists.

In practical implementations, numerically stable methods such as QR decomposition or SVD are generally preferred over explicitly calculating the matrix inverse.

---

# Gradient Descent

Gradient Descent is another optimization method that can be used to minimize the cost function.

The parameters are updated repeatedly:

$$
\boxed{
\beta_j
\leftarrow
\beta_j
-
\alpha
\frac{\partial J}{\partial \beta_j}
}
$$

Where:

* $\beta_j$ = Model parameter
* $\alpha$ = Learning Rate
* $J$ = Cost Function

---

# Learning Rate

The Learning Rate determines the size of each parameter update.

$$
\boxed{
\alpha = \text{Learning Rate}
}
$$

* Very small $\alpha$ → Slow training
* Very large $\alpha$ → May overshoot the minimum
* Appropriate $\alpha$ → Faster and more stable convergence

---

# Multiple Polynomial Regression

When multiple features are present, Polynomial Regression can create polynomial and interaction features.

Suppose there are two features:

$$
x_1,\ x_2
$$

For degree 2, the polynomial features are:

$$
x_1,\ x_2,\ x_1^2,\ x_1x_2,\ x_2^2
$$

The model becomes:

$$
y =
\beta_0
+
\beta_1x_1
+
\beta_2x_2
+
\beta_3x_1^2
+
\beta_4x_1x_2
+
\beta_5x_2^2
$$

---

# Interaction Features

Polynomial Feature transformation can create **interaction terms**.

For example:

$$
x_1x_2
$$

represents an interaction between two features.

Interaction features allow the model to learn situations where the effect of one feature depends on another feature.

---

# Number of Polynomial Features

Polynomial transformation can significantly increase the number of features.

For $p$ original features and degree $d$, when all polynomial terms up to degree $d$ are included:

$$
\boxed{
\text{Number of Features}
=
\binom{p+d}{d}
}
$$

Example:

$$
p=10,\quad d=5
$$

Then:

$$
\binom{15}{5}=3003
$$

Therefore:

$$
\boxed{
\text{Higher Degree}
\Rightarrow
\text{More Features}
}
$$

---

# Overfitting

Polynomial Regression can easily overfit when the degree is too high.

### Low Degree

* Model may be too simple.
* Cannot capture the underlying pattern.
* May cause **underfitting**.

### Appropriate Degree

* Captures the important pattern.
* Generalizes well to unseen data.

### High Degree

* Model becomes highly flexible.
* Starts learning noise.
* Training error becomes very low.
* Test/validation error may increase.
* Causes **overfitting**.

---

# Underfitting vs Overfitting

| Polynomial Degree | Result       |
| ----------------- | ------------ |
| Too Low           | Underfitting |
| Appropriate       | Good Fit     |
| Too High          | Overfitting  |

Generally:

$$
\boxed{
Degree \uparrow
\Rightarrow
Complexity \uparrow
\Rightarrow
Overfitting\ Risk \uparrow
}
$$

---

# Bias-Variance Tradeoff

Polynomial degree affects the bias and variance of the model.

### Low Degree

$$
Bias \uparrow
$$

$$
Variance \downarrow
$$

### High Degree

$$
Bias \downarrow
$$

$$
Variance \uparrow
$$

The goal is to find a balance between bias and variance.

---

# Choosing the Polynomial Degree

The polynomial degree should not be selected only using training error.

Use:

* Validation Set
* Cross-Validation
* Validation MSE
* Validation RMSE
* Test Set for final evaluation

Example:

| Degree | Validation MSE |
| -----: | -------------: |
|      1 |             25 |
|      2 |             12 |
|      3 |              7 |
|      4 |              9 |
|      5 |             18 |

In this example:

$$
\boxed{
\text{Best Degree} = 3
}
$$

because it has the lowest validation MSE.

---

# Feature Scaling

Polynomial features can have very different magnitudes.

For example:

$$
x=100
$$

$$
x^2=10,000
$$

$$
x^3=1,000,000
$$

Therefore, feature scaling can be useful, especially when using:

* Gradient Descent
* Ridge Regression
* Lasso Regression
* ElasticNet

### Standardization

$$
\boxed{
x' = \frac{x-\mu}{\sigma}
}
$$

Where:

* $\mu$ = Mean
* $\sigma$ = Standard Deviation

---

# Polynomial Regression with Regularization

Polynomial features can create many features and increase the risk of overfitting.

Regularization can be used to control model complexity.

## Polynomial + Ridge

$$
\boxed{
\text{Polynomial Features} + \text{Ridge}
}
$$

* Uses L2 regularization.
* Shrinks coefficients.
* Helps reduce overfitting.

---

## Polynomial + Lasso

$$
\boxed{
\text{Polynomial Features} + \text{Lasso}
}
$$

* Uses L1 regularization.
* Can make some coefficients exactly zero.
* Can perform feature selection.

---

## Polynomial + ElasticNet

$$
\boxed{
\text{Polynomial Features} + \text{ElasticNet}
}
$$

* Uses both L1 and L2 regularization.
* Shrinks coefficients.
* Can perform feature selection.

---

# Assumptions of Polynomial Regression

Polynomial Regression using ordinary least squares generally relies on assumptions such as:

1. **Linearity in Parameters:** The model should be linear with respect to its coefficients.

2. **Independence:** Observations should generally be independent.

3. **Homoscedasticity:** Error variance should remain approximately constant.

4. **Zero Conditional Mean:** Errors should have approximately zero mean given the predictors.

5. **No Severe Multicollinearity:** Polynomial features should not create problematic correlations.

Polynomial features such as:

$$
x,\ x^2,\ x^3
$$

can become highly correlated with each other.

---

# Model Evaluation

Common evaluation metrics include:

* $R^2$ Score
* Mean Squared Error (MSE)
* Root Mean Squared Error (RMSE)
* Mean Absolute Error (MAE)

---

# R² Score

R² Score measures how much of the variance in the target variable is explained by the model.

$$
\boxed{
R^2 =
1 -
\frac{
\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
}{
\sum_{i=1}^{n}(y_i-\bar{y})^2
}
}
$$

Where:

* $y_i$ = Actual value
* $\hat{y}_i$ = Predicted value
* $\bar{y}$ = Mean of actual values

---

# Interpretation of R² Score

| R² Score | General Interpretation                  |
| -------: | --------------------------------------- |
|      1.0 | Perfect fit                             |
|      0.9 | Very strong fit                         |
|      0.8 | Good fit                                |
|      0.6 | Moderate fit                            |
|      0.0 | No improvement over predicting the mean |
|      < 0 | Worse than predicting the mean          |

> **Higher R² generally indicates better fit, but it should not be used alone to evaluate the model.**

---

# Advantages of Polynomial Regression

* Can model non-linear relationships.
* More flexible than Linear Regression.
* Simple to understand and implement.
* Uses Linear Regression algorithms internally.
* Can capture feature interactions.
* Useful when the relationship follows a polynomial pattern.

---

# Disadvantages of Polynomial Regression

* High degree can cause overfitting.
* Number of features can increase rapidly.
* Can introduce multicollinearity.
* High powers can create numerical instability.
* Sensitive to outliers.
* Degree selection is important.
* Not every non-linear relationship can be represented well using a polynomial.

---

# Polynomial Regression vs Linear Regression

| Feature                | Linear Regression    | Polynomial Regression       |
| ---------------------- | -------------------- | --------------------------- |
| Relationship           | Linear               | Non-linear / Curved         |
| Equation               | $y=\beta_0+\beta_1x$ | $y=\beta_0+\beta_1x+\cdots$ |
| Feature Transformation | No                   | Yes                         |
| Degree                 | 1                    | 2, 3, 4, ...                |
| Flexibility            | Low                  | Higher                      |
| Overfitting Risk       | Lower                | Higher                      |
| Uses Linear Regression | —                    | Yes                         |

---

# Key Takeaways

* Polynomial Regression is used for **curved/non-linear relationships**.
* It extends Linear Regression by adding polynomial features.
* General equation:

$$
\boxed{
\hat{y}
=
\beta_0
+
\beta_1x
+
\beta_2x^2
+
\cdots
+

}
$$

* Polynomial Regression can be viewed as:

$$
\boxed{
\text{Polynomial Features}
+

}
$$

* Polynomial degree controls model complexity.

* High degree can cause **overfitting**.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures

from sklearn.datasets import fetch_california_housing
from sklearn.linear_model import LinearRegression

from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score
)
```

## Dataset

```python
housing = fetch_california_housing(as_frame=True)

df = housing.frame
```

```python
X_train, X_test, y_train, y_test = train_test_split(
  X, y, test_size=0.20, random_state=42
)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
degree = 2

model = Pipeline([
    
    # Create polynomial features
    (
        "polynomial_features",
        PolynomialFeatures(
            degree=degree,
            include_bias=False
        )
    ),
    
    # Scale features
    (
        "scaler",
        StandardScaler()
    ),
    
    # Linear Regression
    (
        "linear_regression",
        LinearRegression()
    )
])
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred,
    "Residual": y_test.values - y_pred
})

print("\nActual vs Predicted:")
print(comparison.head(10))
```

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test,y_pred)
```

```python
print(f"MAE  : {mae:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.5
)

plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs Predicted Values")
plt.show()
```

```python
residuals = y_test - y_pred
plt.figure(figsize=(8, 6))
plt.scatter(
    y_pred,
    residuals,
    alpha=0.5
)
plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Predicted Values")
plt.ylabel("Residuals")
plt.title("Residual Plot")
plt.show()
```

```python
degrees = [1, 2, 3]

results = []

for degree in degrees:

    model_degree = Pipeline([

        (
            "polynomial_features",
            PolynomialFeatures(
                degree=degree,
                include_bias=False
            )
        ),

        (
            "scaler",
            StandardScaler()
        ),

        (
            "linear_regression",
            LinearRegression()
        )
    ])

    model_degree.fit(
        X_train,
        y_train
    )

    predictions = model_degree.predict(
        X_test
    )

    mse_value = mean_squared_error(
        y_test,
        predictions
    )

    rmse_value = np.sqrt(mse_value)

    r2_value = r2_score(
        y_test,
        predictions
    )

    results.append({

        "Degree": degree,

        "MSE": mse_value,

        "RMSE": rmse_value,

        "R2": r2_value

    })
```

```python
results_df = pd.DataFrame(results)

print("\nPolynomial Degree Comparison:")

print(results_df)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    results_df["Degree"],
    results_df["RMSE"],
    marker="o"
)

plt.xlabel("Polynomial Degree")

plt.ylabel("RMSE")

plt.title("Polynomial Degree vs RMSE")

plt.xticks(degrees)

plt.show()
```
