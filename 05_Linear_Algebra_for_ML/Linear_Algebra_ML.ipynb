# Linear Algebra for Machine Learning (Part 1)

## Introduction

Linear Algebra is the mathematical foundation of Machine Learning. Almost every ML algorithm represents data as vectors and matrices and performs mathematical operations on them.

It is widely used in:

- Linear Regression
- Logistic Regression
- Neural Networks
- PCA
- Recommendation Systems
- Computer Vision
- Natural Language Processing

# Why Linear Algebra?

Machine Learning models cannot understand tables or text directly. They convert everything into numbers.

Example:

| Age | Salary |
|------|---------|
|20|25000|
|25|35000|
|30|50000|

becomes

\[
X=
\begin{bmatrix}
20 & 25000\\
25 & 35000\\
30 & 50000
\end{bmatrix}
\]

Every ML algorithm performs operations on this matrix.

# Matrix Multiplication

## Definition

Matrix multiplication combines two matrices to produce a new transformed matrix.

---

## Mathematical Rule

If

\[
A_{m\times n}
\]

and

\[
B_{n\times p}
\]

then

\[
AB=C_{m\times p}
\]

The inner dimensions must be equal.

---

## Formula

Each element is computed as

\[
c_{ij}=\sum_{k=1}^{n}a_{ik}b_{kj}
\]

---

## Example

\[
A=
\begin{bmatrix}
1&2\\
3&4
\end{bmatrix}
\]

\[
B=
\begin{bmatrix}
5&6\\
7&8
\end{bmatrix}
\]

Result

\[
AB=
\begin{bmatrix}
19&22\\
43&50
\end{bmatrix}
\]

---

## Mathematical Intuition

A matrix represents a transformation.

Matrix multiplication applies that transformation to another matrix or vector.

It can:

- Rotate
- Scale
- Translate
- Transform

the data.

---

## ML Intuition

Every Dense Layer in a Neural Network computes

\[
Y=WX+b
\]

where

- \(X\) = Input
- \(W\) = Weight Matrix
- \(b\) = Bias
- \(Y\) = Output

This operation is simply matrix multiplication.

---

## ML Applications

- Linear Regression
- Logistic Regression
- Neural Networks
- CNN
- Transformers
- PCA

# Dot Product

## Definition

The Dot Product (Inner Product) is an operation between two vectors that produces a single scalar value.

It measures how similar or aligned two vectors are.

---

## Mathematical Formula

For two vectors

\[
A =
\begin{bmatrix}
a_1\\
a_2\\
\vdots\\
a_n
\end{bmatrix}
,\qquad
B =
\begin{bmatrix}
b_1\\
b_2\\
\vdots\\
b_n
\end{bmatrix}
\]

The dot product is

\[
A \cdot B = \sum_{i=1}^{n} a_i b_i
\]

or

\[
A \cdot B = a_1b_1+a_2b_2+\cdots+a_nb_n
\]

---

## Example

\[
A=
\begin{bmatrix}
1\\
2\\
3
\end{bmatrix}
,
B=
\begin{bmatrix}
4\\
5\\
6
\end{bmatrix}
\]

\[
A\cdot B=(1\times4)+(2\times5)+(3\times6)=32
\]

---

## Mathematical Intuition

The dot product measures how much one vector points in the same direction as another.

It can also be written as

\[
A\cdot B=||A||||B||\cos\theta
\]

where

- \(||A||\) = Magnitude of vector A
- \(||B||\) = Magnitude of vector B
- \(\theta\) = Angle between vectors

If

- \(\theta=0^\circ\) → Maximum similarity
- \(\theta=90^\circ\) → No similarity
- \(\theta=180^\circ\) → Opposite direction

---

## ML Applications

- Linear Regression
- Logistic Regression
- Neural Networks
- Cosine Similarity
- Recommendation Systems
- Word Embeddings

---

# Cross Product

## Definition

The Cross Product is an operation between two 3D vectors that produces another vector.

Unlike the dot product, the result is a vector.

---

## Mathematical Formula

For vectors

\[
A=(a_1,a_2,a_3)
\]

\[
B=(b_1,b_2,b_3)
\]

\[
A\times B=
\begin{bmatrix}
a_2b_3-a_3b_2\\
a_3b_1-a_1b_3\\
a_1b_2-a_2b_1
\end{bmatrix}
\]

---

## Example

\[
A=
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix}
,
B=
\begin{bmatrix}
0\\
1\\
0
\end{bmatrix}
\]

\[
A\times B=
\begin{bmatrix}
0\\
0\\
1
\end{bmatrix}
\]

---

## Mathematical Intuition

The resulting vector is

- Perpendicular to both vectors.
- Magnitude equals the area of the parallelogram formed by the vectors.

---

## ML Applications

Rarely used in traditional ML.

Common in

- Robotics
- Computer Vision
- 3D Graphics
- Autonomous Vehicles

---

# Transpose

## Definition

Transpose converts rows into columns and columns into rows.

---

## Mathematical Representation

If

\[
A=
\begin{bmatrix}
1&2&3\\
4&5&6
\end{bmatrix}
\]

Then

\[
A^T=
\begin{bmatrix}
1&4\\
2&5\\
3&6
\end{bmatrix}
\]

---

## Mathematical Intuition

Transpose changes the orientation of data without changing its values.

Rows become columns.

Columns become rows.

---

## ML Applications

- Matrix Multiplication
- Covariance Matrix
- Linear Regression
- PCA
- Gradient Computation

---

# Determinant

## Definition

The determinant is a scalar value computed from a square matrix.

It indicates whether a matrix is invertible.

---

## Formula (2×2 Matrix)

For

\[
A=
\begin{bmatrix}
a&b\\
c&d
\end{bmatrix}
\]

\[
|A|=ad-bc
\]

---

## Example

\[
A=
\begin{bmatrix}
1&2\\
3&4
\end{bmatrix}
\]

\[
|A|=(1)(4)-(2)(3)
=-2
\]

---

## Mathematical Intuition

The determinant measures the scaling factor of a transformation.

- Positive → Orientation preserved
- Negative → Orientation flipped
- Zero → Matrix collapses into lower dimensions

---

## Interpretation

| Determinant | Meaning |
|-------------|---------|
| > 0 | Valid transformation |
| < 0 | Reflection occurred |
| = 0 | Matrix has no inverse |

---

## ML Applications

- Matrix Inversion
- PCA
- Gaussian Distribution
- Linear Regression

---

# Rank

## Definition

Rank is the number of linearly independent rows or columns in a matrix.

---

## Mathematical Representation

If

\[
A=
\begin{bmatrix}
1&2\\
2&4
\end{bmatrix}
\]

Row 2 is twice Row 1.

Therefore,

\[
Rank(A)=1
\]

---

## Example

\[
A=
\begin{bmatrix}
1&2\\
3&4
\end{bmatrix}
\]

Both rows are independent.

\[
Rank(A)=2
\]

---

## Mathematical Intuition

Rank tells us how much unique information exists in the matrix.

Higher rank means

- More independent features
- More information

Lower rank means

- Redundant information
- Duplicate features

---

## ML Applications

- Feature Selection
- Detecting Redundant Features
- PCA
- Dimensionality Reduction
- Data Compression

# Inverse Matrix

## Definition

The inverse of a matrix is a matrix that reverses the effect of the original matrix.

It is similar to division in normal arithmetic.

If

\[
A^{-1}
\]

is the inverse of matrix \(A\), then

\[
AA^{-1}=A^{-1}A=I
\]

where \(I\) is the Identity Matrix.

---

## Formula (2×2 Matrix)

For

\[
A=
\begin{bmatrix}
a&b\\
c&d
\end{bmatrix}
\]

The inverse is

\[
A^{-1}
=
\frac{1}{ad-bc}
\begin{bmatrix}
d&-b\\
-c&a
\end{bmatrix}
\]

where

\[
ad-bc \neq 0
\]

---

## Mathematical Intuition

The inverse "undoes" a matrix transformation.

Example:

- Addition is reversed by subtraction.
- Multiplication is reversed by division.
- Matrix transformation is reversed by the inverse matrix.

---

## Conditions

A matrix has an inverse only if:

- It is a square matrix.
- Its determinant is not zero.

---

## ML Applications

- Linear Regression (Normal Equation)
- Solving Linear Equations
- Optimization
- Matrix Computations

---

# Eigenvalues

## Definition

An Eigenvalue tells us how much a vector is stretched or compressed after a transformation.

It is represented by

\[
\lambda
\]

---

## Mathematical Formula

\[
Av=\lambda v
\]

where

- \(A\) = Matrix
- \(v\) = Eigenvector
- \(\lambda\) = Eigenvalue

---

## Mathematical Intuition

Imagine stretching a rubber sheet.

Some directions stretch more than others.

The amount of stretching is the **Eigenvalue**.

- Large Eigenvalue → More important direction
- Small Eigenvalue → Less important direction

---

## ML Applications

- Principal Component Analysis (PCA)
- Image Compression
- Spectral Clustering
- Face Recognition

---

# Eigenvectors

## Definition

An Eigenvector is a special vector whose direction remains unchanged after a matrix transformation.

Only its magnitude changes.

---

## Mathematical Formula

\[
Av=\lambda v
\]

---

## Mathematical Intuition

Normally, multiplying a vector by a matrix changes both its length and direction.

For an Eigenvector:

- Direction remains the same.
- Only the length changes.

---

## ML Applications

- PCA
- Feature Extraction
- Data Compression
- Face Recognition

---

# Relationship Between Eigenvalues and Eigenvectors

The equation

\[
Av=\lambda v
\]

means

- \(v\) gives the important direction.
- \(\lambda\) tells how important that direction is.

Together, they identify the principal directions of the data.

---

# Norms

## Definition

A Norm measures the length (magnitude) of a vector.

---

## L1 Norm (Manhattan Norm)

Formula

\[
||x||_1=\sum_{i=1}^{n}|x_i|
\]

Example

\[
[2,-3,5]
\]

\[
||x||_1=2+3+5=10
\]

---

## L2 Norm (Euclidean Norm)

Formula

\[
||x||_2=\sqrt{\sum_{i=1}^{n}x_i^2}
\]

Example

\[
[3,4]
\]

\[
||x||_2=5
\]

---

## Infinity Norm

Formula

\[
||x||_\infty=\max(|x_i|)
\]

Example

\[
[2,-10,5]
\]

\[
||x||_\infty=10
\]

---

## Mathematical Intuition

Norm tells how "large" a vector is.

Different norms measure size in different ways.

---

## ML Applications

- Regularization
- Distance Measurement
- Optimization
- Gradient Descent

---

# Orthogonality

## Definition

Two vectors are orthogonal if they are perpendicular.

Their dot product is zero.

---

## Mathematical Formula

For vectors

\[
A
\]

and

\[
B
\]

If

\[
A\cdot B=0
\]

then

\[
A\perp B
\]

---

## Example

\[
A=
\begin{bmatrix}
1\\
0
\end{bmatrix}
,
B=
\begin{bmatrix}
0\\
1
\end{bmatrix}
\]

\[
A\cdot B=0
\]

Hence, the vectors are orthogonal.

---

## Mathematical Intuition

Orthogonal vectors are completely independent.

Knowing one vector gives no information about the other.

---

## ML Applications

- PCA
- QR Decomposition
- Feature Engineering
- Dimensionality Reduction

---

# Singular Value Decomposition (SVD)

## Definition

Singular Value Decomposition (SVD) is a matrix factorization technique that decomposes any matrix into three simpler matrices.

---

## Mathematical Formula

For any matrix

\[
A
\]

\[
A=U\Sigma V^T
\]

where

- \(U\) = Left Singular Vectors
- \(\Sigma\) = Singular Values
- \(V^T\) = Right Singular Vectors

---

## Mathematical Intuition

Instead of working with one large complex matrix, SVD breaks it into three simpler parts.

It identifies the most important patterns while removing redundant information.

---

## Why is SVD Useful?

SVD helps to:

- Compress data
- Reduce dimensions
- Remove noise
- Find hidden patterns

---

## ML Applications

- Principal Component Analysis (PCA)
- Recommendation Systems (Netflix, Amazon)
- Image Compression
- Natural Language Processing (LSA)
- Noise Reduction

```python
M = np.array([
    [3,1],
    [1,3]
])

U,S,VT = np.linalg.svd(M)

print("Matrix")
print(M)

print("\nU Matrix")
print(U)

print("\nSingular Values")
print(S)

print("\nV Transpose")
print(VT)
```

```python
X = np.array([
    [20,170,65],
    [22,175,70],
    [25,180,75]
])

print("Dataset")
print(X)

print("\nTranspose")
print(X.T)

print("\nFeature Means")
print(np.mean(X,axis=0))

print("\nFeature Standard Deviation")
print(np.std(X,axis=0))

print("\nCovariance Matrix")
cov = np.cov(X.T)
print(cov)

print("\nEigenvalues of Covariance Matrix")
values,vectors = np.linalg.eig(cov)
print(values)

print("\nEigenvectors")
print(vectors)

print("\nPrincipal Direction (Largest Eigenvalue)")
print(vectors[:,np.argmax(values)])
```
